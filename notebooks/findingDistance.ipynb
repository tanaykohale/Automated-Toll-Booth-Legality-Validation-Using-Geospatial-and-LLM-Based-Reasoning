```python

from math import radians, sin, cos, sqrt, asin


EARTH_RADIUS_KM = 6371.0
DISTANCE_THRESHOLD_KM = 60

#  Haversine distance between 2 points
def haversine(lat1, lon1, lat2, lon2):
    # convert to radians
    lat1, lon1, lat2, lon2 = map(radians, [lat1, lon1, lat2, lon2])
    dlat = lat2 - lat1
    dlon = lon2 - lon1
    a = sin(dlat / 2)**2 + cos(lat1) * cos(lat2) * sin(dlon / 2)**2
    c = 2 * asin(sqrt(a))
    return EARTH_RADIUS_KM * c

#Main Function
def get_close_pairs(df, distance_km=DISTANCE_THRESHOLD_KM):
    pairs = []
    n = len(df)
    
    for i in range(n):
        id1, lat1, lon1 = df.iloc[i][['id', 'lat', 'lon']]
        for j in range(i + 1, n):  # avoid duplicate and self-pairs
            id2, lat2, lon2 = df.iloc[j][['id', 'lat', 'lon']]
            dist = haversine(lat1, lon1, lat2, lon2)
            if dist <= distance_km:
                pairs.append({
                    'id_1': id1, 'id_2': id2,
                    'lat1': lat1, 'lon1': lon1,
                    'lat2': lat2, 'lon2': lon2
                })
    
    return pd.DataFrame(pairs)
# Get pairs
pairs_df = get_close_pairs(df)
```
